```python
# First, install sklearn if you don't have it
# !pip install scikit-learn pandas

# Import the tools we need
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import CategoricalNB #(NaiveBayes)
from sklearn.metrics import accuracy_score
```

```python

# Create our tiny Play Tennis dataset
data = {
    'Weather': ['Sunny', 'Sunny', 'Overcast', 'Rainy', 'Rainy', 'Rainy', 'Overcast', 'Sunny', 'Sunny', 'Rainy', 'Sunny', 'Overcast', 'Overcast', 'Rainy'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool', 'Cool', 'Mild', 'Cool', 'Mild', 'Mild', 'Mild', 'Hot', 'Mild'],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)
```

```python
df.head(10)
```

```python

# Change text to numbers because Naive Bayes in sklearn likes numbers
df_encoded = df.copy()
#Save original dataframe, work safely on a copy
```

# why we are using .cat.codes and not get_dummies() here?

```python
#Turn text into integer codes (good for CategoricalNB)

df_encoded['Weather'] = df_encoded['Weather'].astype('category').cat.codes
df_encoded['Temperature'] = df_encoded['Temperature'].astype('category').cat.codes
df_encoded['Play'] = df_encoded['Play'].astype('category').cat.codes
```

```python
df_encoded.head(10) #Sunny ->2 , Overcast->0 , Rainy -->1
#Hot-->1,Mild-->2, Cool-->0 
#Yes-->1, No--->0
```

```python
#why we are not using get_dummies here and instead we using cat.codes?
```

```python

# Split into input (X) and output (y)
X = df_encoded[['Weather', 'Temperature']]
y = df_encoded['Play']
```

```python
X.head()
```

```python
y.head()
```

```python

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

```

```python
# Make the model
model = CategoricalNB() #NaiveBayes

# Train (teach) the model
model.fit(X_train, y_train)

# What the model learns when you call .fit(X_train, y_train):

# Basic Class Probabilities:
# P(Play = Yes)
# P(Play = No)

# Feature: Weather
# P(Weather = Sunny | Play = Yes)
# P(Weather = Overcast | Play = Yes)
# P(Weather = Rainy | Play = Yes)
# P(Weather = Sunny | Play = No)
# P(Weather = Overcast | Play = No)
# P(Weather = Rainy | Play = No)

# Feature: Temperature
# P(Temperature = Hot | Play = Yes)
# P(Temperature = Mild | Play = Yes)
# P(Temperature = Cool | Play = Yes)
# P(Temperature = Hot | Play = No)
# P(Temperature = Mild | Play = No)
# P(Temperature = Cool | Play = No)


```

```python

# Test (ask) the model
y_pred = model.predict(X_test)
```

```python

# How good did it do?
print("Accuracy:", accuracy_score(y_test, y_pred)*100,"%")
```

```python
 #Sunny ->2 , Overcast->0 , Rainy -->1
#Hot-->1,Mild-->2, Cool-->0 
#Yes-->1, No--->0
```
